```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data_frame=pd.read_csv('data.csv')
data_frame.head()
```

```python
data_frame.shape
```

```python
data_frame=data_frame.drop(columns=['id','Unnamed: 32'])
data_frame.head()
```

```python
data_frame.info()
```

```python
data_frame.describe()
```

```python
data_frame.isnull().sum()
```

```python
data_frame.duplicated().sum()
```

```python
data_frame['diagnosis'].value_counts()
```

```python
data_frame["diagnosis"] = data_frame["diagnosis"].map({"M": 1, "B": 0})
```

```python
data_frame['diagnosis'].head()
```

```python
data_frame.diagnosis.value_counts()
```

```python
y=data_frame['diagnosis']
y.head()
```

```python
X=data_frame.drop(columns=['diagnosis'])
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
```

```python
print(X_train.shape,y_train.shape)
```

```python
print(X_test.shape,y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(X_train)
X_train_standardized=scaler.transform(X_train)
X_test_standardized=scaler.transform(X_test)

```

```python
X_train_standardized.mean(axis=0)
```

```python
X_train_standardized.std(axis=0)
```

```python
X_train_standardized=np.asarray(X_train_standardized)
X_test_standardized= np.asarray(X_test_standardized)

y_train=np.asarray(y_train).reshape(-1,1)
y_test=np.asarray(y_test).reshape(-1,1)
```

```python
X_train_standardized.shape,y_train.shape
```

```python
X_test_standardized.shape,y_test.shape
```

```python
np.savez("Processed_data.npz",X_train=X_train_standardized,X_test=X_test_standardized,y_train=y_train,y_test=y_test)

```
